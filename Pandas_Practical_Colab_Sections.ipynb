## Setup – Import libraries and create the sample dataset

Run this cell first. It creates `df`, which is used by Parts 4–12.

```python
# Setup: Import libraries and create the sample dataset

# ============================================================
# PANDAS DATA ANALYSIS – PART 4 TO PART 12
# Complete Google Colab Practical Program
# ============================================================

import pandas as pd
import numpy as np

print("Pandas Version:", pd.__version__)

# ============================================================
# CREATE SAMPLE DATASET
# ============================================================

data = {
    "OrderID": [101, 102, 103, 104, 105, 106, 107, 108,
                109, 110, 111, 112, 113, 113],
    "Name": ["Rahul", "Priya", "Amit", "Sneha", "Kiran",
             "Ravi", "Anjali", "Arun", "Meena", "John",
             "Divya", "Raj", "Neha", "Neha"],
    "Department": ["Sales", "HR", "IT", "Sales", "IT",
                   "HR", "Sales", "IT", "HR", "Sales",
                   "IT", "HR", "Sales", "Sales"],
    "City": ["Bangalore", "Mumbai", "Delhi", "Bangalore",
             "Chennai", "Mumbai", "Delhi", "Chennai",
             "Bangalore", "Mumbai", "Delhi", "Chennai",
             "Bangalore", "Bangalore"],
    "Quantity": [5, 3, 8, 4, 6, 2, 7, 9, 3, 5, 4, 6, 100, 100],
    "UnitPrice": [200, 300, 150, 250, 400, 350, 180,
                  220, 500, 280, 320, 150, 2000, 2000],
    "Sales": [1000, 900, 1200, 1000, 2400, 700, 1260,
              1980, 1500, 1400, 1280, 900, 200000, 200000],
    "OrderDate": [
        "2025-01-05", "2025-01-10", "2025-02-15",
        "2025-02-20", "2025-03-05", "2025-03-12",
        "2025-04-10", "2025-04-15", "2025-05-05",
        "2025-05-12", "2025-06-10", "2025-06-15",
        "2025-07-10", "2025-07-10"
    ],
    "Email": [
        "RAHUL@gmail.com", "PRIYA@gmail.com",
        "AMIT@gmail.com", "SNEHA@gmail.com",
        "KIRAN@gmail.com", "RAVI@gmail.com",
        "ANJALI@gmail.com", "ARUN@gmail.com",
        "MEENA@gmail.com", "JOHN@gmail.com",
        "DIVYA@gmail.com", "RAJ@gmail.com",
        "NEHA@gmail.com", "NEHA@gmail.com"
    ],
    "Score": [85, 90, 78, 88, 92, 75, 95, 80, 89,
              70, 94, 82, np.nan, np.nan]
}

df = pd.DataFrame(data)
df.loc[2, "City"] = np.nan
df.loc[5, "Quantity"] = np.nan
df.loc[8, "UnitPrice"] = np.nan
df["OrderDate"] = pd.to_datetime(df["OrderDate"])

print("\\nORIGINAL DATASET")
display(df)
print("\\nDATASET INFORMATION")
df.info()
print("\\nFIRST FIVE ROWS")
display(df.head())
```

## Part 4 – Selecting and Filtering Data

```python
# PART 4: SELECTING AND FILTERING DATA

print("\\n========== PART 4: SELECTING AND FILTERING ==========")

# Selecting a Single Column
print("\\nSingle Column:")
display(df["Name"])

# Selecting Multiple Columns
print("\\nMultiple Columns:")
display(df[["Name", "Department", "Sales"]])

# loc[] - Select rows and columns using labels
print("\\nloc[]:")
display(df.loc[0:4, ["Name", "Sales"]])

# iloc[] - Select rows and columns using positions
print("\\niloc[]:")
display(df.iloc[0:5, 0:4])

# Conditional Filtering
print("\\nSales Greater Than 1000:")
display(df[df["Sales"] > 1000])

# Multiple Conditions - AND
print("\\nAND Condition:")
display(df[(df["Department"] == "Sales") & (df["Sales"] > 1000)])

# Multiple Conditions - OR
print("\\nOR Condition:")
display(df[(df["Department"] == "HR") | (df["Department"] == "IT")])

# NOT Condition
print("\\nNOT Condition:")
display(df[~(df["Department"] == "Sales")])

# isin()
print("\\nisin():")
display(df[df["Department"].isin(["Sales", "IT"])])

# between()
print("\\nbetween():")
display(df[df["Sales"].between(900, 2000)])

# query()
print("\\nquery():")
display(df.query("Sales > 1000 and Quantity > 3"))

# filter()
print("\\nfilter():")
display(df.filter(items=["Name", "Sales", "City"]))

# select_dtypes()
print("\\nNumeric Columns:")
display(df.select_dtypes(include=np.number))
print("\\nObject Columns:")
display(df.select_dtypes(include="object"))
```

## Part 5 – Data Cleaning and Preprocessing

```python
# PART 5: DATA CLEANING AND PREPROCESSING

print("\\n========== PART 5: DATA CLEANING ==========")

clean_df = df.copy()

# isnull() and isna()
print("\\nisnull():")
display(clean_df.isnull())
print("\\nisna():")
display(clean_df.isna())

# notnull()
print("\\nnotnull():")
display(clean_df.notnull())

# dropna()
print("\\ndropna():")
display(clean_df.dropna())

# fillna()
print("\\nfillna():")
clean_df["City"] = clean_df["City"].fillna("Unknown")
clean_df["Quantity"] = clean_df["Quantity"].fillna(clean_df["Quantity"].mean())
clean_df["UnitPrice"] = clean_df["UnitPrice"].fillna(clean_df["UnitPrice"].mean())
clean_df["Score"] = clean_df["Score"].fillna(clean_df["Score"].mean())
display(clean_df)

# Forward Fill and Backward Fill
print("\\nForward Fill:")
display(df.ffill())
print("\\nBackward Fill:")
display(df.bfill())

# duplicated() and drop_duplicates()
print("\\nduplicated():")
display(df.duplicated())
print("\\ndrop_duplicates():")
clean_df = clean_df.drop_duplicates()
display(clean_df)

# replace()
print("\\nreplace():")
clean_df["Department"] = clean_df["Department"].replace({"Sales": "Marketing"})
display(clean_df["Department"])

# astype()
print("\\nastype():")
clean_df["OrderID"] = clean_df["OrderID"].astype(str)
print(clean_df.dtypes)

# convert_dtypes()
print("\\nconvert_dtypes():")
print(clean_df.convert_dtypes().dtypes)

# rename()
print("\\nrename():")
clean_df = clean_df.rename(columns={"Name": "CustomerName", "Sales": "TotalSales"})
display(clean_df.head())

# drop()
print("\\ndrop():")
display(df.drop(columns=["Email"]).head())

# Handling Outliers - IQR Method
print("\\nIQR OUTLIER DETECTION:")
Q1 = clean_df["TotalSales"].quantile(0.25)
Q3 = clean_df["TotalSales"].quantile(0.75)
IQR = Q3 - Q1
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = clean_df[
    (clean_df["TotalSales"] < lower_limit) |
    (clean_df["TotalSales"] > upper_limit)
]
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Outliers:")
display(outliers)

clean_no_outliers = clean_df[
    (clean_df["TotalSales"] >= lower_limit) &
    (clean_df["TotalSales"] <= upper_limit)
]
print("\\nDataset After Removing Outliers:")
display(clean_no_outliers)
```

## Part 6 – Data Manipulation and Transformation

```python
# PART 6: DATA MANIPULATION AND TRANSFORMATION

print("\\n========== PART 6: DATA MANIPULATION ==========")
df2 = df.copy()

# Adding a New Column
df2["TotalAmount"] = df2["Quantity"] * df2["UnitPrice"]
print("\\nNew Column:")
display(df2[["Name", "Quantity", "UnitPrice", "TotalAmount"]])

# Updating Existing Columns
df2["Department"] = df2["Department"].replace("HR", "Human Resources")
print("\\nUpdated Department:")
display(df2["Department"])

# Deleting Columns
print("\\nDeleted Email Column:")
display(df2.drop(columns=["Email"]).head())

# insert()
df2.insert(2, "Country", "India")
print("\\ninsert():")
display(df2.head())

# apply()
df2["SalesTax"] = df2["Sales"].apply(lambda x: x * 0.18)
print("\\napply():")
display(df2[["Sales", "SalesTax"]])

# map()
df2["DepartmentCode"] = df2["Department"].map({
    "Sales": 1, "HR": 2, "IT": 3
})
print("\\nmap():")
display(df2[["Department", "DepartmentCode"]])

# DataFrame.map()
print("\\nDataFrame.map():")
small_df = pd.DataFrame({"A": [1, 2, 3], "B": [4, 5, 6]})
display(small_df.map(lambda x: x * 2))

# assign()
print("\\nassign():")
assigned_df = df.assign(
    Discount=lambda x: x["Sales"] * 0.10,
    FinalAmount=lambda x: x["Sales"] * 0.90
)
display(assigned_df[["Sales", "Discount", "FinalAmount"]])

# where() and mask()
print("\\nwhere():")
display(df["Sales"].where(df["Sales"] > 1000, 0))
print("\\nmask():")
display(df["Sales"].mask(df["Sales"] > 1000, 0))

# Lambda Functions
print("\\nLambda Function:")
df2["PriceCategory"] = df2["Sales"].apply(
    lambda x: "High" if x > 1500 else "Low"
)
display(df2[["Sales", "PriceCategory"]])

# Vectorized Operations
print("\\nVectorized Operations:")
df2["DoubleSales"] = df2["Sales"] * 2
df2["Profit"] = df2["Sales"] - (df2["Sales"] * 0.20)
display(df2[["Sales", "DoubleSales", "Profit"]])
```

## Part 7 – Sorting and Ranking

```python
# PART 7: SORTING AND RANKING

print("\\n========== PART 7: SORTING AND RANKING ==========")

# sort_values()
print("\\nsort_values():")
display(df.sort_values(by="Sales", ascending=False))

# sort_index()
print("\\nsort_index():")
display(df.sort_index(ascending=False))

# rank()
print("\\nrank():")
df_rank = df.copy()
df_rank["SalesRank"] = df_rank["Sales"].rank(ascending=False, method="dense")
display(df_rank[["Name", "Sales", "SalesRank"]])

# nlargest() and nsmallest()
print("\\nnlargest():")
display(df.nlargest(5, "Sales"))
print("\\nnsmallest():")
display(df.nsmallest(5, "Sales"))
```

## Part 8 – GroupBy and Aggregation

```python
# PART 8: GROUPBY AND AGGREGATION

print("\\n========== PART 8: GROUPBY AND AGGREGATION ==========")

# Understanding groupby() / Grouping by One Column
print("\\ngroupby() - Total Sales by Department:")
display(df.groupby("Department")["Sales"].sum())

print("\\nAverage Sales by Department:")
display(df.groupby("Department")["Sales"].mean())

# Grouping by Multiple Columns
print("\\nGrouping by Department and City:")
display(df.groupby(["Department", "City"], dropna=False)["Sales"].sum())

# Aggregation Functions and agg()
print("\\nAggregation Functions:")
display(df.groupby("Department")["Sales"].agg(["sum", "mean", "min", "max", "count"]))

print("\\nagg():")
display(df.groupby("Department").agg({
    "Sales": ["sum", "mean"],
    "Quantity": ["sum", "max"],
    "Score": ["mean", "min"]
}))

# Named Aggregation
print("\\nNamed Aggregation:")
display(df.groupby("Department").agg(
    Total_Sales=("Sales", "sum"),
    Average_Sales=("Sales", "mean"),
    Total_Quantity=("Quantity", "sum")
))

# transform()
print("\\ntransform():")
df_transform = df.copy()
df_transform["DepartmentAverage"] = (
    df_transform.groupby("Department")["Sales"].transform("mean")
)
display(df_transform[["Name", "Department", "Sales", "DepartmentAverage"]])

# GroupBy filter()
print("\\nGroupBy filter():")
filtered_groups = df.groupby("Department").filter(
    lambda x: x["Sales"].mean() > 1000
)
display(filtered_groups)

# Custom Aggregation
print("\\nCustom Aggregation:")
def sales_range(x):
    return x.max() - x.min()

display(df.groupby("Department")["Sales"].agg(sales_range))

# reset_index()
print("\\nreset_index():")
grouped = df.groupby("Department")["Sales"].sum()
display(grouped.reset_index())
```

## Part 9 – Combining Datasets

```python
# PART 9: COMBINING DATASETS

print("\\n========== PART 9: COMBINING DATASETS ==========")

data2 = {
    "OrderID": [101, 102, 103, 104, 105, 106],
    "PaymentMode": ["Cash", "UPI", "Card", "UPI", "Cash", "Card"],
    "Status": ["Paid", "Paid", "Pending", "Paid", "Paid", "Pending"]
}
df_payment = pd.DataFrame(data2)

# concat()
print("\\nconcat():")
concat_df = pd.concat([df.head(3), df.iloc[3:6]], ignore_index=True)
display(concat_df)

# merge() and Join Types
print("\\nmerge() - Inner Join:")
display(pd.merge(df, df_payment, on="OrderID", how="inner"))

print("\\nLeft Join:")
display(pd.merge(df, df_payment, on="OrderID", how="left"))

print("\\nRight Join:")
display(pd.merge(df, df_payment, on="OrderID", how="right"))

print("\\nOuter Join:")
display(pd.merge(df, df_payment, on="OrderID", how="outer"))

# Merging Using Multiple Columns
print("\\nMerging Multiple Columns:")
left_data = pd.DataFrame({
    "Department": ["Sales", "IT", "HR"],
    "City": ["Delhi", "Mumbai", "Chennai"],
    "Employees": [10, 15, 8]
})
right_data = pd.DataFrame({
    "Department": ["Sales", "IT", "HR"],
    "City": ["Delhi", "Mumbai", "Chennai"],
    "Budget": [50000, 80000, 30000]
})
display(pd.merge(left_data, right_data, on=["Department", "City"], how="inner"))

# Self Join
print("\\nSelf Join:")
employee = pd.DataFrame({
    "EmployeeID": [1, 2, 3, 4],
    "EmployeeName": ["Amit", "Priya", "Rahul", "Sneha"],
    "ManagerID": [np.nan, 1, 1, 2]
})
self_join = pd.merge(
    employee, employee,
    left_on="ManagerID", right_on="EmployeeID",
    how="left", suffixes=("", "_Manager")
)
display(self_join[["EmployeeName", "EmployeeName_Manager"]].rename(
    columns={"EmployeeName": "Employee", "EmployeeName_Manager": "Manager"}
))

# join()
print("\\njoin():")
df_left = pd.DataFrame({"Name": ["Amit", "Priya", "Rahul"]}, index=[1, 2, 3])
df_right = pd.DataFrame({"Salary": [25000, 30000, 35000]}, index=[1, 2, 3])
display(df_left.join(df_right))
```

## Part 10 – Date and Time Operations

```python
# PART 10: DATE AND TIME OPERATIONS

print("\\n========== PART 10: DATE AND TIME ==========")
date_df = df.copy()
date_df["OrderDate"] = pd.to_datetime(date_df["OrderDate"])

# Extracting Year, Month, Day and Weekday
date_df["Year"] = date_df["OrderDate"].dt.year
date_df["Month"] = date_df["OrderDate"].dt.month
date_df["Day"] = date_df["OrderDate"].dt.day
date_df["Weekday"] = date_df["OrderDate"].dt.day_name()

print("\\nYear, Month, Day and Weekday:")
display(date_df[["OrderDate", "Year", "Month", "Day", "Weekday"]])

# Date Filtering
print("\\nDate Filtering:")
display(date_df[
    (date_df["OrderDate"] >= "2025-03-01") &
    (date_df["OrderDate"] <= "2025-05-31")
])

# Date Differences
date_df["DaysDifference"] = (
    pd.Timestamp("2025-12-31") - date_df["OrderDate"]
).dt.days
print("\\nDate Differences:")
display(date_df[["OrderDate", "DaysDifference"]])

# date_range()
print("\\ndate_range():")
dates = pd.date_range(start="2025-01-01", end="2025-01-10", freq="D")
display(dates)

# Time Zones
print("\\nTime Zones:")
utc_dates = pd.Series(pd.date_range("2025-01-01", periods=3, freq="D", tz="UTC"))
india_dates = utc_dates.dt.tz_convert("Asia/Kolkata")
print("UTC:")
display(utc_dates)
print("India Time:")
display(india_dates)

# resample()
print("\\nresample():")
monthly_data = (
    date_df.set_index("OrderDate")["Sales"]
    .resample("ME")
    .sum()
)
display(monthly_data)

# rolling()
print("\\nrolling():")
date_df = date_df.sort_values("OrderDate")
date_df["RollingAverage"] = date_df["Sales"].rolling(window=3, min_periods=1).mean()
display(date_df[["OrderDate", "Sales", "RollingAverage"]])

# shift()
print("\\nshift():")
date_df["PreviousSales"] = date_df["Sales"].shift(1)
display(date_df[["Sales", "PreviousSales"]])
```

## Part 11 – String Operations

```python
# PART 11: STRING OPERATIONS

print("\\n========== PART 11: STRING OPERATIONS ==========")
string_df = df.copy()

# str.lower(), str.upper(), str.title()
print("\\nstr.lower():")
display(string_df["Name"].str.lower())

print("\\nstr.upper():")
display(string_df["Name"].str.upper())

print("\\nstr.title():")
display(string_df["Email"].str.title())

# str.strip()
print("\\nstr.strip():")
sample_text = pd.Series(["  Rahul  ", " Priya ", " Amit  "])
display(sample_text.str.strip())

# str.replace()
print("\\nstr.replace():")
display(string_df["Email"].str.replace("gmail.com", "outlook.com", regex=False))

# str.contains()
print("\\nstr.contains():")
display(string_df[
    string_df["Email"].str.contains("rahul", case=False, na=False)
])

# str.startswith() and str.endswith()
print("\\nstr.startswith():")
display(string_df[string_df["Name"].str.startswith("A")])

print("\\nstr.endswith():")
display(string_df[string_df["Email"].str.endswith("gmail.com", na=False)])

# str.split()
print("\\nstr.split():")
display(string_df["Email"].str.split("@"))

# str.extract()
print("\\nstr.extract():")
display(string_df["Email"].str.extract(r"@(.+)$"))

# str.len()
print("\\nstr.len():")
display(string_df["Name"].str.len())

# Regular Expressions (Regex)
print("\\nRegular Expressions:")
display(string_df[
    string_df["Name"].str.contains(r"^[AR]", regex=True, na=False)
])

text_data = pd.Series(["Order101", "Order102", "Order103"])
print("\\nExtract numbers using Regex:")
display(text_data.str.extract(r"(\d+)"))
```

## Part 12 – Statistical Analysis

```python
# PART 12: STATISTICAL ANALYSIS

print("\\n========== PART 12: STATISTICAL ANALYSIS ==========")
sales_data = df["Sales"]

# mean()
print("\\nMean:")
print(sales_data.mean())

# median()
print("\\nMedian:")
print(sales_data.median())

# mode()
print("\\nMode:")
display(sales_data.mode())

# sum() and count()
print("\\nSum:")
print(sales_data.sum())

print("\\nCount:")
print(sales_data.count())

# min() and max()
print("\\nMinimum:")
print(sales_data.min())

print("\\nMaximum:")
print(sales_data.max())

# std() and var()
print("\\nStandard Deviation:")
print(sales_data.std())

print("\\nVariance:")
print(sales_data.var())

# quantile()
print("\\nQuantile:")
display(sales_data.quantile([0.25, 0.50, 0.75]))

# corr() and cov()
print("\\nCorrelation:")
display(df[["Quantity", "UnitPrice", "Sales", "Score"]].corr())

print("\\nCovariance:")
display(df[["Quantity", "UnitPrice", "Sales", "Score"]].cov())

# cumsum()
print("\\nCumulative Sum:")
display(sales_data.cumsum())

# cumprod()
print("\\nCumulative Product:")
display(pd.Series([2, 3, 4, 5]).cumprod())

# pct_change()
print("\\nPercentage Change:")
display(sales_data.pct_change() * 100)

# diff()
print("\\nDifference:")
display(sales_data.diff())

# ============================================================
# FINAL OUTPUT
# ============================================================

print("\\n==========================================")
print("ALL PANDAS OPERATIONS COMPLETED SUCCESSFULLY")
print("PART 4 TO PART 12")
print("==========================================")

print("\\nFinal Dataset:")
display(df.head(10))

print("\\nDataset Shape:")
print(df.shape)

print("\\nDataset Description:")
display(df.describe(include="all"))
```
